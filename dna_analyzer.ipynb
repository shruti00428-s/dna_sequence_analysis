```python

from Bio import SeqIO

fasta_file = "human_tp53_sequence.fasta"

record = SeqIO.read(fasta_file, "fasta")
sequence = str(record.seq).upper()

valid_bases = set("ACGTN")

if not sequence:
    raise ValueError("The FASTA sequence is empty.")

if not set(sequence).issubset(valid_bases):
    raise ValueError("Unexpected characters in the DNA sequence.")

counts = {
    base: sequence.count(base)
    for base in "ATGCN"
}

length = len(sequence)
gc_content = (counts["G"] + counts["C"]) / length * 100
at_content = (counts["A"] + counts["T"]) / length * 100

print("TP53 Sequence Analysis")
print("FASTA ID:", record.id)
print("Sequence length:", length)
print("Nucleotide counts:", counts)
print(f"GC content: {gc_content:.2f}%")
print(f"AT content: {at_content:.2f}%")
```
